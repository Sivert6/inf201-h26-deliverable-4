## Task 1: Temperatures

`Temperature` stores a value and a unit, which can be Celsius (`"C"`), Fahrenheit (`"F"`) or Kelvin (`"K"`). The unit is case-insensitive and defaults to `"C"`. An invalid unit or a temperature below absolute zero (0 K) raises a `ValueError`. `value` and `unit` are read-only: they are stored as the private attributes `_value` and `_unit` and exposed through `@property` without a setter.

The class has the conversion methods `to_celsius()`, `to_fahrenheit()` and `to_kelvin()`, and `convert_to()` returns a new `Temperature` in the target unit. `__repr__` gives an unambiguous representation such as `Temperature(25.0, 'C')`, while `__str__` gives a readable string with two decimals such as `25.00 °C`. For Kelvin the degree symbol is left out (`273.15 K`), following SI notation.

All six comparisons convert both temperatures to Kelvin first, so temperatures in different units can be compared directly. Since floating-point calculations give small rounding errors, two temperatures are considered equal if they differ by less than 1e-9. Comparing with a non-Temperature returns `NotImplemented`.

```python
body = Temperature(37, "c")
print(body, repr(body))
print(body.convert_to("F"), body.convert_to("K"))
print(Temperature(0, "C") == Temperature(32, "F"), Temperature(0, "C") < Temperature(50, "F"))
print(Temperature(37, "C") == Temperature(98.6, "F"))

try:
    Temperature(-300, "C")
except ValueError as err:
    print(f"ValueError: {err}")

try:
    body.value = 40
except AttributeError as err:
    print(f"AttributeError: {err}")
```

**Result:** The unit `"c"` is converted to `"C"`, so the unit is case-insensitive. `str()` gives the readable `37.00 °C`, while `repr()` gives `Temperature(37.0, 'C')`. Body temperature is converted correctly to 98.60 °F and 310.15 K. The comparisons work across units: 0 °C equals 32 °F, and 0 °C is lower than 50 °F. `37 °C == 98.6 °F` is `True` even though the conversion gives a tiny rounding error, since we use a tolerance of 1e-9. A temperature below absolute zero raises a `ValueError`, and trying to change `value` raises an `AttributeError`, since the property has no setter.

## Task 2: Fractions

`Fraction` stores a fraction with an integer numerator and a strictly positive denominator. Both must be integers, otherwise a `TypeError` is raised, and a denominator of 0 or below raises a `ValueError`. The fraction is always reduced with `math.gcd`, so `Fraction(6, 8)` is stored as `3/4`.

The class implements the initialiser, all six comparisons and six other special methods: `__repr__`, `__str__`, `__add__`, `__sub__`, `__mul__` and `__truediv__`. Two fractions a/b and c/d are compared by cross-multiplying, so we compare a·d with c·b. This works since both denominators are positive. Comparing with a non-Fraction returns `NotImplemented`.

```python
a = Fraction(1, 2)
b = Fraction(3, 4)

print(Fraction(6, 8), repr(Fraction(6, 8)))
print(a + b, a - b, a * b, a / b)
print(a == Fraction(2, 4), a != b, a < b, a <= b, a > b, a >= b)

try:
    Fraction(1, 0)
except ValueError as err:
    print(f"ValueError: {err}")
```

**Result:** `Fraction(6, 8)` is reduced to `3/4`. `str()` gives the readable `3/4`, while `repr()` gives `Fraction(3, 4)`. The arithmetic gives 1/2 + 3/4 = 5/4, 1/2 - 3/4 = -1/4, 1/2 · 3/4 = 3/8 and (1/2) / (3/4) = 2/3, all reduced. `1/2 == 2/4` is `True`, since the comparison uses cross-multiplication. A denominator of 0 raises a `ValueError`.
