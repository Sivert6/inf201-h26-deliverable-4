## Task 2: Fractions

`Fraction` stores a fraction with an integer numerator and a strictly positive denominator. Both must be integers, otherwise a `TypeError` is raised, and a denominator of 0 or below raises a `ValueError`. The fraction is always reduced with `math.gcd`, so `Fraction(6, 8)` is stored as `3/4`.

The class implements the initialiser, all six comparisons and six other special methods: `__repr__`, `__str__`, `__add__`, `__sub__`, `__mul__` and `__truediv__`. Two fractions a/b and c/d are compared by cross-multiplying, so we compare a·d with c·b. This works since both denominators are positive. Comparing with a non-Fraction returns `NotImplemented`.

```python
a = Fraction(1, 2)
b = Fraction(3, 4)

print(Fraction(6, 8), repr(Fraction(6, 8)))
print(a + b, a - b, a * b, a / b)
print(a == Fraction(2, 4), a != b, a < b, a <= b, a > b, a >= b)

try:
    Fraction(1, 0)
except ValueError as err:
    print(f"ValueError: {err}")
```

**Result:** `Fraction(6, 8)` is reduced to `3/4`. `str()` gives the readable `3/4`, while `repr()` gives `Fraction(3, 4)`. The arithmetic gives 1/2 + 3/4 = 5/4, 1/2 - 3/4 = -1/4, 1/2 · 3/4 = 3/8 and (1/2) / (3/4) = 2/3, all reduced. `1/2 == 2/4` is `True`, since the comparison uses cross-multiplication. A denominator of 0 raises a `ValueError`.
